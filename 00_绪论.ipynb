## 1. 环境、图像表示与基础操作
**目标**：能稳定运行 OpenCV-Python，理解图像在计算机中的表示；建立对 OpenCV 库本身的定位认知。

### OpenCV 在图像处理中的地位
- 起源与定位：OpenCV（Open Source Computer Vision Library），C++ 编写、跨平台开源计算机视觉库，由英特尔发起，现在是全球传统机器视觉领域事实标准库。OpenCV-Python 只是它的 Python 绑定封装，核心算法依旧跑C++。
- 能力边界：专注图像/视频底层算法实现，覆盖从像素操作、滤波形态学、特征提取、相机标定到深度学习推理（DNN）；它**不是深度学习训练框架**（不做模型训练），也不是图像可视化库（Matplotlib/Pillow侧重可视化）。
- 行业应用场景：工业质检、安防监控、机器人视觉、自动驾驶感知底层、相机标定、AR标记识别、嵌入式视觉、教学科研；工业界传统视觉方案首选。
- 和同类库对比：
  - PIL/Pillow：侧重图像读写、简单像素处理，没有轮廓、特征、标定等高级视觉算法；
  - scikit-image：Python原生视觉库，API优雅，但性能弱，工业落地较少；
  - PyTorch/TensorFlow：深度学习训练+推理，不擅长传统图像预处理、几何变换、相机几何。
- 优缺点总结：
  - 优势：算法成熟、性能高、跨平台（PC/ARM/嵌入式）、海量案例、社区庞大、商用免费；
  - 局限：高层封装弱，偏向底层算法；Python接口只是封装，纯Python循环容易产生性能瓶颈；不同API版本间存在接口变更。
- 学习定位：OpenCV 是**计算机视觉的工具箱**，学会它不等于掌握全部CV，重点是学会用它实现视觉算法，而不是死记API。

### 知识点清单
- Python、NumPy、OpenCV 安装（pip / conda / 源码编译）
- `opencv-python` 与 `opencv-contrib-python` 的区别
- `opencv-python-headless`（服务器、无 GUI 环境）
- 版本差异与常见坑（OpenCV 3.x / 4.x / 5.x）
- 验证安装：`cv2.__version__`、`cv2.getBuildInformation()`
- 图像的基本表示：`ndarray`、shape、dtype、通道顺序（BGR vs RGB）
- 8位 / 16位 / 32F 图像的差异与适用场景
- 图像内存布局：行优先、连续内存、步长（stride）
- 像素访问、ROI、视图（view）vs 副本（copy）
- 色彩空间初步认知：灰度、BGR、RGB、HSV、YCrCb、Lab
- 图像的读取、显示、保存：`imread`、`imshow`、`imwrite`
- 图像元数据（EXIF）读取基础
- 视频的读取与保存：`VideoCapture`、`VideoWriter`
- 摄像头 / 视频文件 / 图像序列的输入输出
- 视频读取属性：`get(CAP_PROP_*)`、`set(CAP_PROP_*)`
- 通道分离与合并：`split` / `merge`
- 图像算术：`add`、`subtract`、`multiply`、`divide`
- 饱和运算 vs 模运算（OpenCV `add` vs NumPy `+`）
- 位运算：`bitwise_and`、`bitwise_or`、`bitwise_not`、`bitwise_xor`
- 掩码操作：`copyTo` 配合 mask
- 图像混合：`addWeighted`
- 性能基础：向量化、避免 Python 循环、`cv2.setUseOptimized`
- 计时：`cv2.getTickCount`、`time.perf_counter`
- OpenCV 的 `UMat` 与 OpenCL 加速初步

**独立性**：只依赖 NumPy 和 OpenCV 基础 API，是后续所有部分的前置。

## 3. 颜色与色彩空间处理
**目标**：能按颜色维度处理图像，为分割、跟踪、增强打基础。

### 知识点清单
- BGR / RGB：通道顺序、`ndarray` 形状、`split` / `merge`、BGR↔RGB 转换
- BGR ↔ Gray：加权公式 `Y = 0.299R + 0.587G + 0.114B`
- BGR/RGB 常见坑：显示时红蓝互换、`imshow` 与 `plt.imshow` 差异
- HSV：H/S/V 物理含义、OpenCV 数值范围（H∈[0,179]，S/V∈[0,255]）
- HSV：为什么 H 是 180 而不是 360、H 的环形特性、红色跨 0 和 179 的两段阈值
- HSV：为什么适合颜色分割（H 与光照解耦）
- HSV_FULL：H∈[0,255] 变体，精度更高
- YCrCb：Y 亮度、Cr 红色色度、Cb 蓝色色度
- YCrCb：数值范围（Y∈[0,255]，Cr/Cb 中心 128）、与 YUV 的关系
- YCrCb：经典肤色阈值（Cr∈[133,173]，Cb∈[77,127]）
- Lab：L 亮度、a 绿-红、b 蓝-黄、OpenCV 8 位映射
- Lab：感知均匀性、颜色差异度量（ΔE）、L 通道光照归一化
- 色彩空间选型：分割用 HSV、肤色用 YCrCb、色彩校正用 Lab、读写用 BGR
- 色彩空间转换：`cvtColor`
- 转换链限制：必须经过 BGR，不能直接从 HSV 转 Lab
- 颜色范围选取与 `inRange`
- 直方图：计算、绘制、均衡化、自适应均衡化
- CLAHE：`createCLAHE`
- 彩色图像均衡化：按通道分离或只均衡 Y/L 通道
- 2D 直方图：H-S 平面联合直方图
- 直方图反向投影：`calcBackProject`
- 颜色分割完整流程：转 HSV → `inRange` → 形态学去噪 → 轮廓提取
- 肤色检测完整流程：YCrCb + 形态学
- 对比度与亮度调整：`convertScaleAbs`
- Gamma 变换：查表法 `LUT`
- 白平衡基础：灰度世界法、完美反射法
- 颜色迁移基础：Reinhard 方法
- 颜色恒常性初步
- 工程陷阱：BGR/RGB 混用、H 范围映射错误、转换链断裂、HSV_FULL 阈值不通用

**独立性**：可单独学习，也可与第 2、4 部分结合。
