## 1. 环境、图像表示与基础操作
**目标**：能稳定运行 OpenCV-Python，理解图像在计算机中的表示，掌握像素级基础运算。

### OpenCV 在图像处理中的地位

- 起源与定位：OpenCV（Open Source Computer Vision Library），C++ 编写、跨平台开源计算机视觉库，由英特尔发起，现在是全球传统机器视觉领域事实标准库。OpenCV-Python 只是它的 Python 绑定封装，核心算法依旧跑 C++。

- 能力边界：专注图像/视频底层算法实现，覆盖从像素操作、滤波形态学、特征提取到相机标定；它**不是深度学习训练框架**，也不是图像可视化库。

- 行业应用场景：工业质检、安防监控、机器人视觉、自动驾驶感知底层、相机标定、AR 标记识别、嵌入式视觉、教学科研。

- 和同类库对比：
  - PIL/Pillow：侧重图像读写、简单像素处理，没有轮廓、特征、标定等高级视觉算法；
  - scikit-image：Python 原生视觉库，API 优雅，但性能弱，工业落地较少；
  - Matplotlib：侧重可视化，不提供视觉算法。

- 优缺点总结：
  - 优势：算法成熟、性能高、跨平台、海量案例、社区庞大、商用免费；
  - 局限：高层封装弱，偏向底层算法；Python 接口只是封装，纯 Python 循环容易产生性能瓶颈；不同 API 版本间存在接口变更。

- 学习定位：OpenCV 是**计算机视觉的工具箱**，学会它不等于掌握全部 CV，重点是学会用它实现视觉算法。

### 知识点清单

- Python、NumPy、OpenCV 安装（pip / conda / 源码编译）
- `opencv-python` 与 `opencv-contrib-python` 的区别
- `opencv-python-headless`（服务器、无 GUI 环境）
- 版本差异与常见坑（OpenCV 3.x / 4.x / 5.x）
- 验证安装：`cv2.__version__`、`cv2.getBuildInformation()`
- 图像的基本表示：`ndarray`、shape、dtype、通道顺序（BGR vs RGB）
- 8 位 / 16 位 / 32F 图像的差异与适用场景
- **位深度与类型转换**：`convertTo`、dtype 转换、位深提升/降级——唯一归属本模块
- 图像内存布局：行优先、连续内存、步长（stride）
- 像素访问、ROI、视图（view）vs 副本（copy）
- 色彩空间初步认知：灰度、BGR、RGB、HSV、YCrCb、Lab
- 图像的读取、显示、保存：`imread`、`imshow`、`imwrite`（编码参数用法在本模块，编码原理见模块 7）
- 基于 Matplotlib 显示图像（注意 RGB 顺序转换）
- 图像元数据（EXIF）读取基础、EXIF 方向标记
- 视频的读取与保存：`VideoCapture`、`VideoWriter`
- 摄像头 / 视频文件 / 图像序列的输入输出
- 视频读取属性：`get(CAP_PROP_*)`、`set(CAP_PROP_*)`
- 多帧图像的读取与保存
- 图像的创建与复制
- 图像的裁剪与拼接

- 获取与修改像素值
- 快速 LUT 替换像素值
- 图像算术：`add`、`subtract`、`multiply`、`divide`
- 饱和运算 vs 模运算（OpenCV `add` vs NumPy `+`）
- 位运算：`bitwise_and`、`bitwise_or`、`bitwise_not`、`bitwise_xor`
- 掩码操作：`copyTo` 配合 mask
- 图像混合：`addWeighted`
- 图像积分运算：`integral`
- 图像归一化处理：`normalize`
- **图像质量评价指标（客观）**：PSNR、SSIM、MSE、MAE、SNR、ΔE（颜色差）——本模块为唯一归属，其他模块只引用不重复；需讲清各指标的**适用边界**（PSNR/SSIM 对模糊与压缩敏感、对结构变化不敏感；ΔE 用于颜色保真）

- **性能意识与计时工具**：向量化思路、避免 Python 循环、`cv2.setUseOptimized`、`cv2.getTickCount`、`time.perf_counter`——本模块只讲"性能意识 + 计时"，系统优化方法论见模块 15

- **标准数据集与基准索引（总入口）**：BSDS500（边缘）、Middlebury（立体）、CDnet（视频）、DIV2K/Set5（超分）、Kodak（压缩）、TID2013（质量评价）——本模块给出索引与用途，各模块按需引用

## 2. 几何变换与图像配准
**目标**：掌握图像空间位置变换，理解插值与反向映射，能完成配准基础操作。

### 知识点清单

- 齐次坐标与变换矩阵（2×3 仿射、3×3 透视）
- 正向映射 vs 反向映射（为什么 `warpAffine` 用反向映射）
- 图像平移：`warpAffine` + 平移矩阵
- 图像缩放：`resize`、`INTER_*` 插值方法
- 图像旋转：`getRotationMatrix2D` + `warpAffine`
- 图像翻转：`flip`
- 图像斜切：仿射矩阵构造
- 图像投影变换：`getPerspectiveTransform`、`warpPerspective`
- 图像重映射：`remap`
- 插值方法：最近邻、双线性、双三次、Lanczos、面积插值
- 插值选择指南：放大用 CUBIC/LANCZOS，缩小用 AREA
- **传统超分辨率（插值放大）**：双三次放大、Lanczos 放大——唯一归属本模块；**边界**：本模块只讲"插值放大"，不含去模糊/去噪的超分；边缘导向插值等增强型超分见模块 12 的复原范畴
- 图像配准基础（基于强度）：`findTransformECC`
- **基于相位相关的配准**：`phaseCorrelate`、亚像素级平移估计、工业对齐场景——唯一归属本模块
- **单应矩阵数学**：单应矩阵的数学推导与求解——唯一归属本模块；基于特征匹配的拼接流程见模块 11；标定校正拼接见模块 14
- **多模态配准的概念定位**：基于互信息的配准（医学影像常用）只做概念介绍，明确"本大纲不展开实现"——唯一归属本模块的概念定位

- 实现"证件照透视校正"：手动选四点 + `getPerspectiveTransform` + `warpPerspective`；
- 用 `findTransformECC` 做两帧图像对齐，并对比不同插值方法对配准精度的影响。

## 3. 色彩空间与颜色处理
**目标**：能按颜色维度处理图像，为分割、增强打基础。

### 知识点清单

- BGR / RGB：通道顺序、`ndarray` 形状、`split` / `merge`、BGR↔RGB 转换
- BGR ↔ Gray：加权公式 `Y = 0.299R + 0.587G + 0.114B`
- BGR/RGB 常见坑：显示时红蓝互换、`imshow` 与 `plt.imshow` 差异
- HSV：H/S/V 物理含义、OpenCV 数值范围（H∈[0,179]，S/V∈[0,255]）
- HSV：为什么 H 是 180 而不是 360、H 的环形特性、红色跨 0 和 179 的两段阈值
- HSV：为什么适合颜色分割（H 与光照解耦）
- HSV_FULL：H∈[0,255] 变体，精度更高
- YCrCb：Y 亮度、Cr 红色色度、Cb 蓝色色度
- YCrCb：数值范围（Y∈[0,255]，Cr/Cb 中心 128）、与 YUV 的关系
- YCrCb：经典肤色阈值（Cr∈[133,173]，Cb∈[77,127]）
- Lab：L 亮度、a 绿-红、b 蓝-黄、OpenCV 8 位映射
- Lab：感知均匀性、颜色差异度量（ΔE 见模块 1）、L 通道光照归一化
- 色彩空间选型：分割用 HSV、肤色用 YCrCb、色彩校正用 Lab、读写用 BGR
- 色彩空间转换：`cvtColor`
- 转换链限制：必须经过 BGR，不能直接从 HSV 转 Lab
- **颜色范围选取与阈值掩码**：`inRange`、阈值区间设计——`inRange` 唯一归属本模块，模块 5 只写"见模块 3"
- 灰度图像的伪彩色处理：`applyColorMap`
- 多模态数据合成的伪彩色图像
- 图像的色彩风格滤镜
- 调节图像的色彩平衡
- 颜色分割完整流程：转 HSV → 颜色阈值掩码 → 形态学去噪 → 轮廓提取
- 肤色检测完整流程：YCrCb + 形态学
- 白平衡基础：灰度世界法、完美反射法
- 颜色迁移基础：Reinhard 方法
- 工程陷阱：BGR/RGB 混用、H 范围映射错误、转换链断裂、HSV_FULL 阈值不通用

### 实践项目
- 做一个实时颜色追踪器：鼠标点选像素 → 自动估计 HSV 阈值 → `inRange` + 形态学 + 外接圆；
- 用 YCrCb 肤色阈值实现"手部区域检测"，并对比不同光照下的稳定性；
- 用 Lab 的 L 通道做光照归一化，提升颜色分割鲁棒性。

**独立性**：可单独学习，也可与模块 4、5 结合。
> **注意**：直方图、Gamma、对比度、CLAHE 等内容已归入模块 4，本模块不再重复；颜色范围阈值掩码工具 `inRange` 唯一归属本模块。

## 5. 图像分割
**目标**：能从图像中分离目标区域，掌握传统分割方法。

## 附 B：Top 20 常见误区速查

| # | 误区 | 正确做法 | 相关模块 |
|---|---|---|---|
| 1 | 用 `plt.imshow` 直接显示 `cv2.imread` 的图，红蓝互换 | 先 BGR→RGB 再显示 | 1、3 |
| 2 | 用 NumPy `+` 做图像加法，像素溢出回绕 | 用 `cv2.add` 饱和运算 | 1 |
| 3 | 认为 HSV 的 H 范围是 [0,360] | OpenCV 中 H∈[0,179] | 3 |
| 4 | 红色阈值只写一段 [0,10] | 红色跨 0 和 179，需两段掩码合并 | 3 |
| 5 | 从 HSV 直接转 Lab | 必须经 BGR 中转 | 3 |
| 6 | 对彩色图直接 `equalizeHist` | 转到 YCrCb/Lab 对亮度通道均衡 | 4 |
| 7 | 用 PSNR 判断去噪好坏，忽略结构指标 | 联合 SSIM + 无参考指标 | 1、4 |
| 8 | 孔洞填充只靠闭运算 | 用形态学重建（模块 6） | 5、6 |
| 9 | 结构元一律用矩形 | 按目标形状选十字/椭圆 | 6 |
| 10 | 放大用 AREA、缩小用 CUBIC | 反过来：放大 CUBIC/LANCZOS，缩小 AREA | 2 |
| 11 | 认为 `warpAffine` 是正向映射 | 内部是反向映射 + 插值 | 2 |
| 12 | 频域滤波忘记 `fftshift` 后滤波再反移 | 遵循 变换→中心化→滤波→反中心化→逆变换 | 7 |
| 13 | Sobel 核构造与边缘检测流程混为一谈 | 核构造在模块 6，阈值/NMS/Canny 在模块 8 | 6、8 |
| 14 | 用 `findContours` 返回的轮廓直接当"点集"做几何计算，忽略近似模式 | 明确 `CHAIN_APPROX_*` 影响点数 | 9 |
| 15 | 把 GLCM 当形状描述子 | GLCM 是纹理描述，形状看链码/傅里叶描述符 | 10 |
| 16 | 特征匹配不做比率测试与 RANSAC | KNN + ratio test + `findHomography(RANSAC)` | 11 |
| 17 | 逆滤波直接除频谱，噪声放大 | 用维纳滤波或加正则 | 12 |
| 18 | 背景建模忽略时间中值滤波作为基线 | 先跑时间中值基线再上 MOG2/KNN | 13 |
| 19 | 标定只看重投影误差，不检查畸变系数是否合理 | 联合检查 k1/k2/p1/p2 量级 | 14 |
| 20 | `cv2` 返回共享内存视图，后续原地修改污染原图 | 需要独立副本时显式 `.copy()` | 1、15 |
